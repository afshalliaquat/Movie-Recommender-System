```python
import ast
```

```python
def convert(obj): #obj is string here
  L=[]
  for i in ast.literal_eval(obj): #convert String to list
    L.append(i['name'])
  return L
```

```python
#convert('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
df['genres']= df['genres'].apply(convert)
```

```python
df['keywords'] =df['keywords'].apply(convert)
```

```python
def convert_cast(obj): #obj is string here
  L=[]
  counter = 0
  for i in ast.literal_eval(obj): #convert String to list
    if counter !=3:
        name =  i['name'].replace(" ","")
        L.append(name)
        counter+=1
    else:
      break
  return L
```

```python
df['cast']=df['cast'].apply(convert_cast)
```

```python
def fetch_director(obj): #obj is string here
  L=[]
  for i in ast.literal_eval(obj): #convert String to list
    if i['job']=='Director':
        L.append(i['name'])

  return L
```

```python
df['crew']=df['crew'].apply(fetch_director)
```

```python
new_df = df[['movie_id','title','tags']]
new_df.head(2)
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df.head()
```

*Stemming is done before vectorization to reduce words like "connected", "connecting" to a common root ("connect"), lowering feature count and improving model efficiency.*

```python
from nltk.stem.porter import PorterStemmer
import re

ps = PorterStemmer()

def preprocess(text):
    y = []
    for i in text.split():
      y.append(ps.stem(i))
    return ' '.join(y)

new_df['tags'] = new_df['tags'].apply(preprocess)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000,stop_words='english')
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
  movie_index = new_df[new_df['title'] == movie].index[0]
  distance = similarity[movie_index]
  movie_list = sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])[1:6]

  for i in movie_list:
    print(new_df.iloc[i[0]].title)
```

```python
import joblib

joblib.dump(similarity,'similarity.pkl')
joblib.dump(new_df,'movies.pkl')
```
